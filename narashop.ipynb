```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.read_csv('pedidos_narashop.csv',sep = ';')
df
```

```python
df['valor_total'] = df['quantidade'] * df['valor_unitario']
df
```

```python
df_categoria_fat = df.groupby('categoria')['valor_total'].sum().reset_index()
df_categoria_fat = df_categoria_fat.sort_values(by='valor_total',ascending = False)
df_categoria_fat

```

```python
df_entregue_500 = df.loc[(df['status_pedido'] == 'Entregue') & (df['valor_total'] >= 500)]
df_entregue_500
```

```python
df_produto_valor_vendido = df.groupby('produto')['valor_total'].sum().reset_index().sort_values(by = 'valor_total', ascending = False)

df_produto_valor_vendido.head()
```

```python
array_produto_val_total = np.array(df_produto_valor_vendido['valor_total'])
array_produto_val_total
```

```python
q1 = np.percentile(array_produto_val_total,25)
q3 = np.percentile(array_produto_val_total,75)

```

```python
df_25_mais = df_produto_valor_vendido.loc[df_produto_valor_vendido ['valor_total'] > q3]
df_25_mais
```

```python
df_25_menos = df_produto_valor_vendido.loc[df_produto_valor_vendido['valor_total']< q1 ]
df_25_menos
```

```python
df_fat_vendedor  = df.groupby('vendedor')['valor_total'].sum().reset_index().sort_values(by = 'valor_total', ascending = False)
df_fat_vendedor
```

```python
array_vendedor = np.array(df_fat_vendedor['valor_total'])
array_vendedor
```

```python
q1 = np.percentile(array_vendedor,25)
q3 = np.percentile(array_vendedor,75)
```

```python
print(f'Q1: {q1}')
print(f'Q3: {q3}')
```

```python
df_vendedor_mais = df_fat_vendedor.loc[df_fat_vendedor['valor_total'] > q3]
df_vendedor_mais
```

```python
df_vendedor_menos = df_fat_vendedor.loc[df_fat_vendedor['valor_total'] < q1]
df_vendedor_menos
```

```python
array_med_total = np.array(df['valor_total'])
array_med_total
```

```python
media = np.mean(array_med_total)
mediana = np.median(array_med_total)
distancia = (media - mediana)/mediana
```

```python
print(f'Média: {media:.2f}')
print(f'Mediana: {mediana}')
print(f'Média confiável: {distancia:.2f}%')
```

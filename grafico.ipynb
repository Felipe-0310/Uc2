```python
import matplotlib.pyplot as plt
import pandas as pd
```

## Dicionário e DataFrame

```python
dados = {'categorias':['Eletrônicos', 'Roupas', 'Alimentos', 'Casa'],
         'Valores': [4200, 2800, 3600, 1900] }
df = pd.DataFrame(dados)
df
```

```python
df = pd.read_csv('vendas_narastore.csv')
df
```

```python
df_categoria_venda = df.groupby('categoria')['total_venda'].sum().reset_index().sort_values(by = 'total_venda',ascending =False)
df_categoria_venda
```

## Grafico Vertical 'bar'

```python
plt.bar(df_categoria_venda['categoria'],df_categoria_venda['total_venda'])
plt.title('Total de Vendas por Categoria — NaraStore')
plt.show()
```

## Gráfico Horizontal 'barH'

```python
plt.barh(df_categoria_venda['categoria'],df_categoria_venda['total_venda'])
plt.title('Total de Vendas por Categoria — NaraStore')
plt.show()
```

## Titulo do eixo X e Y

```python
plt.bar(df_categoria_venda['categoria'],df_categoria_venda['total_venda'])
plt.title('Total de Vendas por Categoria — NaraStore')
plt.xlabel('Categorias')
plt.ylabel('Quantidade')
plt.show()
```

## Rotação dos dados

```python
plt.bar(df_categoria_venda['categoria'],df_categoria_venda['total_venda'])
plt.title('Total de Vendas por Categoria — NaraStore')
plt.xticks(rotation = 45)
plt.show()
```

## Aumentar as informações(largurax,alturay)

```python
plt.figure(figsize = (12,7))
plt.bar(df_categoria_venda['categoria'],df_categoria_venda['total_venda'])
plt.title('Total de Vendas por Categoria — NaraStore')
plt.show()
```
